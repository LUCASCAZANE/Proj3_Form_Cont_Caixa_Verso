# Conjunto de dados de classificação de aprovação de empréstimos
#### Dados sintéticos para classificação binária na aprovação de empréstimos

## Base de Dados:
### https://www.kaggle.com/datasets/taweilo/loan-approval-classification-data/data

# Sobre o conjunto de dados:
### Este conjunto de dados é uma versão sintética inspirada no conjunto de dados original de Risco de Crédito do Kaggle e enriquecida com variáveis ​​adicionais baseadas em dados de Risco Financeiro para Aprovação de Empréstimos . O SMOTENC foi usado para simular novos pontos de dados e ampliar as instâncias. O conjunto de dados está estruturado para recursos categóricos e contínuos.

# Objetivo do Projeto:
### Classificação : Construir modelos preditivos para classificar a loan_statusvariável (aprovado/não aprovado) para potenciais candidatos.
